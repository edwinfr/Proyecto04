# Contador de palabras

Este cuaderno pide una frase al usuario y luego muestra cuántas palabras contiene.

```python
import re

frase = input("Ingresa una frase: ")
palabras = re.findall(r"\b\w+\b", frase.lower())
print(f"La frase tiene {len(palabras)} palabras")
```
